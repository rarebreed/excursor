## Why Bayesian Networks?

Standard feed-forward networks have several limitations:

1. They are prone to overfitting, 
2. They use point estimates, instead of distributions
3. Use maximum likelihood estimation, 

One may ask why the above are limitations.  We will explain each of these limitations, and how Bayesian networks address them. 

### Overfitting

Overfitting occurs when a model does not generalize well to new data. It essentially memorizes the training data, instead of learning the underlying patterns.  It will capture noise and outliers in the training data, leading to poor performance on new data. 

Overfitting can be mitigated by using techniques such as regularization, cross-validation, and early stopping. However, Bayesian networks use probability distributions to represent the parameters.  

#### Regularization

Regularization tries to reduce the complexity of a model.  One way it can do this is by making the loss function reduce large weight values, which cause large fluctuations in the model's predictions when presented with new data.  Another way is called dropout, where the model will randomly "turn off" some neurons which forces the model to learn the feature in a different way. 

### Point Estimates

Point estimates are a single value that represents the best estimate of a parameter, but because they are single values, they do not capture the uncertainty of the estimate.  For example, they do not include the mean or variance of the parameter.  This leads to overconfidence in the model's predictions. 

Bayesian networks use probability distributions to represent the parameters.  Since they are distributions, they represent possible values of the parameter and their associated probabilities.  

### Maximum Likelihood Estimation

Related to point estimates is the use of maximum likelihood estimation.  Maximum likelihood estimation basically tries to find the parameters that maximize the likelihood of the data.  This is a common approach in machine learning, but it has the issue of overfitting as mentioned above.
