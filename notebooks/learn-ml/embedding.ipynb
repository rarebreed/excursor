# Text embedding

All models require working with numbers, typically fp numbers.  So we need a way to convert text into numbers.
More specifically a word is represented as a vector of fp numbers of some medium to high dimensionality.

Here we will break down how to create a simple text embedding program

1. tokenize: split text to find words, punctuation, symbols, etc
    - sentence? paragraph?
    - is whitespace important?
2. create token ID for each token
3. mapping from token -> token_id and token_id -> token
4. encoder fn that takes new text, and maps to list of IDs
5. decoder fn that takes a list of IDs, and returns list of tokens
