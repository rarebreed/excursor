## Building a Variational Deep Embedding (VaDE) model

This notebook is a high level view of creating a VaDE to do clustering of test logs.  This will allow
one to group similiar test logs together without reuiring categories defined ahead-of-time, or to create
labelled training data.

The first part is to create a VAE which will give our initial prior.  The steps include:

1. Gathering abd processing test logs
2. Tokenizing and embedding the log data
    - The input and embedding layer
3. Hidden layers to shrink the dimensions
    - last hidden layer is our latent space
4. Latent space layer outputs 2 values per node: mean and variance

Once we have the latent space, we need another model to do the clustering of the data.
